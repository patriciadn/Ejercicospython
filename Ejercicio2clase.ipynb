```python
import math

#a) Semieje mayor
a = 6378137.0
#c) Semieje menor
c = 6356752.314245
#radio de la esfera de comparación
r = 6371000
e2 = 1 - (c**2) / (a**2)
e = math.sqrt(e2)
#Utilizar la fórmula del Área del esferoide oblato
área_esferoide_oblato = 2*math.pi*a**2*(1+(1-e**2)/e*math.atanh(e))
print("Área de la superficie del esferoide oblato:",área_esferoide_oblato,"metros cuadrados")
#Área de la esfera
Esfera=4*math.pi*r**2
print("Área de la superficie de la esfera:",Esfera,"metros cuadrados")
#Diferencia de áreas
Diferencia=área_esferoide_oblato-Esfera
print("Diferencia:",Diferencia,"metros")
#Cociente
Cociente=área_esferoide_oblato/Esfera
print("Cociente:",Cociente,"metros")
diferencia_porceptual=(área_esferoide_oblato-Esfera)/área_esferoide_oblato*100
print("Diferencia porceptual",diferencia_porceptual)
```

Escribe un programa que pida al usuario su edad (entero), y muestre su categoria de acuerdo a las siguientes reglas:

1.- Niño/a: 0 a 11 años

2.- Adolescente: 12 a 17 años

3.- Joven: 18 a 29 años

4.- Adulto: 30 a 64 años

5.- Adulto mayor: 65 o más años

Si el usuario ingresa un número negativo, que ejecute de nuevo el programa.

```python

    a = int(input("Ingresa tu edad: "))
    if a < 0:
        print("Ingresa un número positivo")
    elif 0 <= a <= 11:
        print("Niño/a")
    elif 12 <= a <= 17:
        print("Adolescente")
    elif 18 <= a <= 29:
        print("Joven")
    elif 30 <= a <= 64:
        print("Adulto")
    elif a >= 65:
        print("Adulto mayor")

```
