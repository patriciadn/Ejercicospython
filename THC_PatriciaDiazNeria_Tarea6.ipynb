```python
import numpy as np

##Ejercicio 1. Sumas de matrices.
def suma_matrices(matriz_a, matriz_b):
  return matriz_a+matriz_b

###Ejercicio 2. Multiplicación entrada a entrada de matrices cuadradas.
def multiplicar_entrada_a_entrada(matriz_a, matriz_b):
  return matriz_a*matriz_b
###2.1 El producto de matrices utilizando una función de numpy.
def multiplicar_matrices_numpy(matriz_a, matriz_b):
  return np.dot(matriz_a, matriz_b)

###Ejercicio 3. La traza de una matriz.
def traza_matriz(matriz):
  return np.trace(matriz)

###Ejercicio 4. La transpuesta de una matriz.
def obtener_transpuesta(matriz):
  return np.transpose(matriz)

###Ejercicio 5. Encontrar el máximo y mínimo elemento en una matriz.
def maximo_minimo(matriz):
  maximo = np.max(matriz)
  minimo = np.min(matriz)
  return maximo, minimo

###Ejercio 6. Calcular la suma del cuadrado de los elementos de una matriz.
def suma_cuadrados(matriz):
  return np.sum(matriz**2)

###Ejercicio 7. Dadas las matrices A y B
#Calcular: a) 2B + 3A, b) (B + 2A)**2, c)A**2 + 3B
def calcular_matrices(matriz_a, matriz_b):
  a=2*matriz_b+3*matriz_a
  c1=(matriz_b+2*matriz_a)
  b=np.dot(c1,c1)
  c=np.dot(matriz_a,matriz_a)+3*matriz_b
  return a,b,c
```

```python
###Ejercicio 7. y ejemplo de matrices.
matriz_a=np.array([[11,7,-1],[-3,5,2],[4,2,-2]])
matriz_b=np.array([[1,1,1],[10,3,-5],[-6,-10,-3]])
print("matriz_a:\n",matriz_a)
print("matriz_b:\n",matriz_b)

print("1) Suma de matrices:\n", suma_matrices(matriz_a, matriz_b))
print("2) El producto de matrices entrada por entrada:\n", multiplicar_entrada_a_entrada(matriz_a, matriz_b))
print("2.1) El producto de matrices utilizando una función de numpy:\n", multiplicar_matrices_numpy(matriz_a, matriz_b))
print("3) La traza de una matriz:\n", traza_matriz(matriz_a))
print("4) La transpuesta de una matriz:\n", obtener_transpuesta(matriz_a))
maximo,minimo=maximo_minimo(matriz_a)
print("5) Encontrar el máximo y mínimo elemento en una matriz:",)
print("Máximo:",int(maximo))
print("Mínimo:",int(minimo))
print("6) Calcular la suma del cuadrado de los elementos de una matriz:\n", suma_cuadrados(matriz_a))
resultado_a,resultado_b,resultado_c=calcular_matrices(matriz_a, matriz_b)
print("7a) Calcular: a) 2B + 3A=\n", resultado_a)
print("\n7b) Calcular:b) (B + 2A)**2=\n",resultado_b)
print("\n 7c) Calcular:c)A**2 + 3B=\n", resultado_c)
```

***###Ejercicio 8. Genere una función que reciba como entradas las dimensiones de una matriz T[n, m] (3 <= n, m <=5). Asegúrese de validar que sea cuadrada; de lo contrario, debe devolver el mensaje de error La matriz no es cuadrada y vuelva a preguntar las entradas. Posteriormente reemplace los valores con números aleatorios***



---


*   La primera fila de manera inversa.
*   La primera columna de manera inversa.
*   El primer elemento y el  ́ultimo elemento.
*   El valor m ́aximo, m ́ınimo, el promedio de la diagonal y la traza.
*   Los valores de T desde T[1,1] hasta T[5,5] (Dependiendo de las dimensiones de la matriz).
*   Los valores desde T[1,5] hasta T[5,1] (Dependiendo de las dimensiones de la matriz).
*   La suma de los elementos pares recorriendo las filas.
*   La suma de los elementos impares recorriendo las columas.
*   Imprima los promedios de las filas y columnas.
*   La antidiagonal y el promedio de sus valores.


```python
import numpy as np

def matriz_T():
    ## Validar entradas
    while True:
        n = int(input("Ingrese el valor de filas n (con 3<=n<=5): "))
        m = int(input("Ingrese el valor de columnas m (con 3<=m<=5): "))

        if n < 3 or n > 5 or m < 3 or m > 5:
            print("n y m deben estar entre 3 y 5. Intente de nuevo.\n")
            continue

        if n != m:
            print("La matriz no es cuadrada. Intente de nuevo.\n")
            continue

        # si pasa las dos validaciones, salimos del while
        break

    ### Crear matriz T con números aleatorios, entre 0 y 15
    T = np.random.randint(0, 15, size=(n, m))
    print("\nMatriz T generada aleatoriamente:")
    print(T)

    #### La primera fila de manera inversa.
    primera_fila = T[0, ::-1]
    print("\nLa primera fila de manera inversa:")
    print(primera_fila)

    #### La primera columna de manera inversa.
    primera_columna = T[::-1, 0]
    print("\nLa primera columna de manera inversa:")
    print(primera_columna)

    #### El primer elemento y el último elemento.
    primer_elemento = T[0, 0]
    ultimo_elemento = T[-1, -1]
    print("\nEl primer elemento y el último elemento:")
    print(primer_elemento)
    print(ultimo_elemento)

    #### El valor máximo, mínimo, el promedio de la diagonal y la traza.
    maximo = np.max(T)
    minimo = np.min(T)
    promedio_diagonal = np.mean(np.diag(T))
    traza = np.trace(T)
    print("\nValor máximo de T:", maximo)
    print("Valor mínimo de T:", minimo)
    print("Promedio de la diagonal de T:", promedio_diagonal)
    print("Traza de T:", traza)

    ## Los valores de T desde T[1,1] hasta T[5,5] (dependiendo de n).
    lim = min(5, n)  # por si n<5
    sub_1_1_a_5_5 = T[0:lim, 0:lim]
    print("\nLos valores de T desde T[1,1] hasta T[5,5]:")
    print(sub_1_1_a_5_5)

    ## Los valores desde T[1,5] hasta T[5,1] (antidiagonal de la misma submatriz). tomando la misma submatriz cuadrada y su antidiagonal
    sub = T[0:lim, 0:lim]
    anti_1_5_a_5_1 = np.fliplr(sub).diagonal()
    print("\nLos valores desde T[1,5] hasta T[5,1]:")
    print(anti_1_5_a_5_1)

    ## La suma de los elementos pares recorriendo las filas.
    suma_pares = 0
    for i in range(n):      # recorriendo filas
        for j in range(m):  # recorriendo columnas
            if T[i, j] % 2 == 0:
                suma_pares += T[i, j]
    print("\nLa suma de los elementos pares recorriendo las filas es:", suma_pares)

    ##### La suma de los elementos impares recorriendo las columnas.
    suma_impares = 0
    for j in range(m):      # recorriendo columnas
        for i in range(n):  # recorriendo filas
            if T[i, j] % 2 != 0:
                suma_impares += T[i, j]
    print("La suma de los elementos impares recorriendo las columnas es:", suma_impares)

    ## Imprima los promedios de las filas y columnas.
    promedios_filas = T.mean(axis=1)
    promedios_columnas = T.mean(axis=0)

    print("\nPromedios de las filas:", promedios_filas)
    print("Promedios de las columnas:", promedios_columnas)

    ## La antidiagonal y el promedio de sus valores.
    antidiagonal = np.fliplr(T).diagonal()
    promedio_antidiagonal = antidiagonal.mean()
    print("\nLa antidiagonal de T es:", antidiagonal)
    print("El promedio de sus valores es:", promedio_antidiagonal)

# probarr
matriz_T()
```
