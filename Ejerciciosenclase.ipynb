***Ejercicios con división entera:***

```python
Manzanas=50
Personas=6
Repartir=Manzanas//Personas
print(Repartir)

Faltan=Manzanas%Personas
print("sobran",Faltan)
```

```python
Dias=100
Dias_por_semana=7
Semanas_totales=Dias//Dias_por_semana
print(Semanas_totales)

Semanas_totales_restantes=Dias%Dias_por_semana
print("sobran",Semanas_totales_restantes)
```

```python
# El Área de un triánguloutilizando la fórmula de Herón esta dado por la siguiente fórmula en Python:
# Longitudes (a), (b), (c)
a=7
b=8
c=9
Semiperimetro=int((a+b+c)/2)
print(Semiperimetro)

Area=(Semiperimetro*(Semiperimetro-a)*(Semiperimetro-b)*(Semiperimetro-c))**0.5
print(Area)
```

```python
#Encontrar las raices con la fórmula general.
#Coeficiente cuadratico (a), Coeficiente lineal (b), Termino independiente (c)
# Coeficiente cuadratico:1
# Coeficiente lineal:-5
# Termino independiente:6
a=int(input("Coeficiente cuadratico: "))
b=int(input("Coeficiente lineal:"))
c=int(input("Termino independiente: "))

x1=int((-b+(b**2-4*a*c)**0.5)/(2*a))
x2=int((-b-(b**2-4*a*c)**0.5)/(2*a))

print(x1)
print(x2)
```
