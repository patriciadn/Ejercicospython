Programe una función que calcule el área de un triángulo utilizando la fórmula de Herón.
Complemente el ejercicio con las condiciones solicitadas en la tarea 3:
1.   Que los valores del triángulo sean positivos. El programa debe repetirse hasta que los tres valores sean positivos.** (hint: ciclo while y condicional if-elif-else).**
2.   Que se cumpla la desigualdad triangular.
3.  Operaciones **(calculo del semiperímetro y el área).**
4. Indique con comentarios sus variables globales y locales.
5.  Llamada a la función con sus argumentos e impresión en pantalla del resultado.





```python
# Variables locales: a, b, c (longitudes de los lados del triángulo), semiperimetro, area
# Variables globales: ninguna
def heron(a, b, c):
  semiperimetro=(a+b+c)/2
  area=(semiperimetro*(semiperimetro-a)*(semiperimetro-b)*(semiperimetro-c))**0.5
  return area

print("Este programa calcula el área de un triángulo utilizando la fórmula de Herón")
while True:
      a = float(input("Ingrese el valor de a: "))
      b = float(input("Ingrese el valor de b: "))
      c = float(input("Ingrese el valor de c: "))
      if a <= 0 or b <= 0 or c <= 0:
        print("Los valores deben ser positivos. Intenta con otros valores")
      elif not (a + b > c and a + c > b and b + c > a):
        print("El triángulo no cumple la desigualdad triangular. Intenta con otros valores")
      else:

        area = heron(a, b, c)
        semiperimetro=(a+b+c)/2
        print(f"El semiperimetro es: {semiperimetro}")
        print(f"El área es {area}")
        break
```
