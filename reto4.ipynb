**2.- Función por intervalos.**

Con Matplotlib, grafique la siguiente función definida por intervalos.

\begin{equation}
f(x)=
\begin{cases}
|x|, & -3 \le x < -1, \\[6pt]
\sin(2x), & -1 \le x < 1, \\[6pt]
e^{0.5x}, & 1 \le x < 4, \\[6pt]
\ln(x-3), & 4 \le x \le 7.
\end{cases}
\end{equation}

Dibuje cada intervalo de diferente color, señale con un legend la función a la que corresponden y posicionelo apropiadamente. Incluya títulos, nombre de los ejes, grid, etc.


```python
import numpy as np
import matplotlib.pyplot as plt

#####intervalos de x
x1 = np.linspace(-3, -1, 200)  # |x|
x2 = np.linspace(-1,  1, 200)  # sin(2x)
x3 = np.linspace( 1,  4, 200)  # e^{0.5x}
x4 = np.linspace( 4,  7, 200)  # ln(x-3)
#####funciones en cada intervalo
y1 = np.abs(x1)
y2 = np.sin(2*x2)
y3 = np.exp(0.5*x3)
y4 = np.log(x4 - 3)
plt.figure(figsize=(10,6))
plt.plot(x1, y1, label=r'$|x|, -3\leq x<-1$', color='turquoise')
plt.plot(x2, y2, label=r'$\sin(2x), -1\leq x<1$', color='magenta')
plt.plot(x3, y3, label=r'$e^{0.5x}, 1\leq x<4$', color='green')
plt.plot(x4, y4, label=r'$\ln(x-3), 4\leq x\leq 7$', color='purple')
# ##### puntos de unión
plt.scatter([-3,-1,1,4,7],
            [np.abs(-3), np.abs(-1),
             np.sin(2*1), np.exp(0.5*4), np.log(7-3)],
            color='black', s=30)
plt.title('Función definida por intervalos', color="skyblue", fontsize=14)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend(loc='best')
plt.show()
```

**3.- Oscilación de funciones.**

En Matplotlib, grafique las siguientes funciones para valores de n=1,2,3,4,5.

\begin{equation}
f_{n}(x)=x^n sin(x)
\end{equation}

\begin{equation}
g_{n}(x)=x^n cos(x)
\end{equation}

Indique con un legend y color la curva para cada valor de **n** y posicione el legend() adecuadamente. Ajuste los límites del eje **y**. O bien, divida entre el máximo de cada función si es necesario por si arroja valores muy grandes. Grafique f(x) y g(x) por separado.

```python
import numpy as np
import matplotlib.pyplot as plt

#####rango de x
x = np.linspace(-10, 10, 4000)
# --- f_n(x) = x^n sin(x) ---
plt.figure(figsize=(9,6))
for n in range(1,6):
    fn = (x**n) * np.sin(x)
    fn_norm = fn / np.max(np.abs(fn))   # dividir entre el máximo
    plt.plot(x, fn_norm, label=fr'$f_{n}(x)=x^{n}\sin(x),\,n={n}$')
plt.title(r'Oscilación de $f_n(x)=x^n\sin(x)$ para $n=1,2,3,4,5$', color="violet")
plt.xlabel('x')
plt.ylabel('f_n(x) normalizada')
plt.grid(True)
plt.legend()
plt.ylim(-1.5, 1.5)
plt.subplots_adjust(bottom=0.35)
plt.show()
# --- g_n(x) = x^n cos(x) ---
plt.figure(figsize=(9,6))
for n in range(1,6):
    gn = (x**n) * np.cos(x)
    gn_norm = gn / np.max(np.abs(gn))
    plt.plot(x, gn_norm, label=fr'$g_{n}(x)=x^{n}\cos(x),\,n={n}$')
plt.title(r'Oscilación de $g_n(x)=x^n\cos(x)$ para $n=1,2,3,4,5$', color="magenta")
plt.xlabel('x')
plt.ylabel('g_n(x) normalizada')
plt.grid(True)
plt.legend()
plt.ylim(-1.5, 1.5)
plt.show()
```
