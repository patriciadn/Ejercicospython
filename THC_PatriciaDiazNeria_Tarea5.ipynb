Ejercicio 3.- Número primo.

Con base en la definición de número primo, escriba una función que reciba un número entero e imprima 'True' si lo es o de lo contrario imprima 'False'. Tenga en cuenta que si el número es menor o igual a 1, se retorna como resultado "Error en el Número" y debe volver a preguntar hasta que la entrada sea valida.

```python
def es_primo(n):
    if n<=1: #Número menor o igual a 1
      print("Error en el número")
      return "Error"
    if n == 2: #Es el único primo par
      return True
    if n % 2 == 0: #Si es par, pero no es 2 entonces no es primo
      return False
    divisor=3
    while divisor*divisor <= n:
      if n % divisor == 0:
        return False
      divisor += 2
    return True

resultado="Error"
while resultado=="Error":
          numero = int(input("Ingrese un número entero: "))
          resultado = es_primo(numero)
          print(resultado)
```

Ejercicio 4.- Números mayores a 100.

Escriba una función que lea un número entero (N) mayor que 100 y valide:

1. Si N es menor de 100 retorne: "Error en el número" y vuelva a preguntar la entrada.
2. Si N termina en 4, 7 ó 9; diga si es divisible por 3.
3. Si N termina en 1 ó 5, divida por 10 y retorne el valor entero,
4. En otro caso divida el número entre 3 y retorne su residuo.

```python
def num_entero(N):
    if N < 100:
        print("Error en el número")
        return "Error"

    # ultimo digito
    ultimo = N % 10

    # ctermina en 4, 7 o 9
    if ultimo in (4, 7, 9):
        if N % 3 == 0:
            return "Es divisible por 3"
        else:
            return "No es divisible por 3"

    # termina en 1 o 5
    elif ultimo in (1, 5):
        return N // 10   # quitar ultimo digito

    # residuo 3
    else:
        return N % 3


resultado = "Error"
while resultado == "Error":
    N = int(input("Ingrese un número entero: "))
    resultado = num_entero(N)
    print(resultado)
```

Ejercicio 5.- MCD y MCM.

Dados dos números a y b, determine:

El máximo común divisor.
El mínimo común múltiplo.
Almacene los resultados en un diccionario e imprima en pantalla el contenido del diccionario.

```python
############# Calcular el maximo comun divisor
def determinar_mcd(a,b):
  while b!=0:
    a,b=b,a%b
  return a
############# Calcular el minimo comun multiplo
def determinar_mcm(a,b):
  mcd=determinar_mcd(a,b)
  return (a*b)//mcd
a=int(input("Ingrese el primer número: "))
b=int(input("Ingrese el segundo número: "))
####Cálculos de mcd y mcm
mcd_resultado = determinar_mcd(a,b)
mcm_resultado = determinar_mcm(a,b)
######Diccionario con los resultados
resultados={"Número A":a,"Número B":b,"MCD":mcd_resultado,"MCM":mcm_resultado}
for clave, valor in resultados.items():
  print(f"{clave}: {valor}")
```

Ejercicio 6.- Filtrado de datos.

Escriba una función que reciba una lista de datos (la de abajo) y devuelva listas con los datos filtrados. Este programa debe separar los datos en int, str, bool, float.

Hint: use la función append para insertar datos en las listas.

```python
datos = [25, "Matemáticas", 3.141592, True, 8, False, "Nombres", 2.718, "UNAM", 0,
         9.81, "CdMx", 7e12, 12.5, "2025", True, "población", 0.01, 42, False]
#Función para separarla lista en sus tipos de datos
def filtrar_datos(lista):
    numeros_enteros=[]
    numeros_decimales=[]
    textos=[]
    val_logicos=[]

    # Recorrer cada elemento de la lista y append para agregar cada elemento a su lista correspondiente
    for elemento in lista:
        if type(elemento)==int and type(elemento)!=bool:   # evitar que True/False se metan como enteros
            numeros_enteros.append(elemento)
        elif type(elemento)==float:
            numeros_decimales.append(elemento)
        elif type(elemento)==str:
            textos.append(elemento)
        elif type(elemento)==bool:
            val_logicos.append(elemento)

    return numeros_enteros, numeros_decimales, textos, val_logicos

# Llamar a la función
numeros_enteros, numeros_decimales, textos, val_logicos = filtrar_datos(datos)

# Resultados
print("Numeros enteros:", numeros_enteros)
print("Numeros decimales:", numeros_decimales)
print("Textos o Cadenas:", textos)
print("Valores lógicos:", val_logicos)
```

**Ejercicio 7.- Cálculo de la distancia entre dos puntos mediante la fórmula de Haversine.**

La fórmula de Haversine permite calcular la distancia entre dos puntos sobre la superficie de la Tierra, a partir de sus coordenadas geográficas. La fórmula está dada por la siguiente expresión:

\begin{equation}
d = 2R  \arcsin \left( \sqrt{\mathrm{haversin}(\varphi_2 - \varphi_1) + \cos(\varphi_1)\cos(\varphi_2)\mathrm{haversin}(\lambda_2 - \lambda_1)} \right)
\end{equation}

donde:

d = distancia entre dos puntos (en km),

R = radio de la Tierra (𝑅≈6378.1),

𝜑1, 𝜑2 = latitudes de los dos puntos.

𝜆1, 𝜆2 = longitudes de los dos puntos (en radianes).

La función de Haversine se define como:

 \begin{equation}
\mathrm{haversin}(\alpha) = \sin^2 \left(\frac{\alpha}{2}\right)
\end{equation}

Para convertir de grados a radianes se utiliza la relación:

\begin{equation}
\text{radianes} = \text{grados} \times \frac{\pi}{180}
\end{equation}

Construya una función que permita calcular la distancia entre dos pares de coordenadas. Utiliza la librería math para las funciones trigonométricas y de raíz cuadrada.

**Ejemplo de prueba.** Coordenadas de Paris y Roma por latitud y longitud: París (48.9, 2.4) y Roma (41.9, 12.5). Resultado: distancia=1107 km.

Proponga 3 pares de coordenadas. Al menos uno debe ser respecto a la Ciudad de México (**Latitud=19.43, Longitud=-99.13**).

```python
import math
R = 6378.1  # Radio de la tierra
#Función Haversine
def haversine(latitud1, long1, latitud2, long2):
    # Convertir de grados a radianes:
    latitud1_rad = math.radians(latitud1)
    long1_rad = math.radians(long1)
    latitud2_rad = math.radians(latitud2)
    long2_rad = math.radians(long2)

    # Fórmula de Haversine
    Hav = math.sin((latitud2_rad - latitud1_rad) / 2)**2 + math.cos(latitud1_rad) * math.cos(latitud2_rad) * math.sin((long2_rad - long1_rad) / 2)**2
    distancia = 2 * R * math.asin(math.sqrt(Hav))

    return distancia

# Ejemplos de pruebas
# París=(48.9, 2.4) Roma=(41.9, 12.5)
distancia_paris_roma = haversine(48.9, 2.4, 41.9, 12.5)
print("Distancia entre París y Roma:", distancia_paris_roma, "km")

# 3 pares de coordenadas
# Suecia=(59.3293, 18.0686)
# Seúl Corea  del Sur (37.5665, 126.9780)
distancia_suecia_seul = haversine(59.3293, 18.0686, 37.5665, 126.9780)
print("Distancia entre Suiza y Corea del Sur:", distancia_suecia_seul, "km")

#-------------------
# Ciudad de México=(19.43, -99.13)
# Italia(Roma)=(41.9028, 12.4964)
distancia_cdmx_roma = haversine(19.43, -99.13, 41.9028, 12.4964)
print("Distancia entre Ciudad de México y Roma:", distancia_cdmx_roma, "km")

#-------------------
# Marruecos(Rabat)=(34.0209, -6.8416)
# Alemania(Berlín)=(52.5200, 13.4050)
distancia_marruecos_alemania = haversine(34.0209, -6.8416, 52.5200, 13.4050)
print("Distancia entre Marruecos y Alemania:", distancia_marruecos_alemania, "km")
```

Ejercicio 8.- La serie de Fibonacci.

Realice un programa que calcule la serie de Fibonacci y la imprima en pantalla, tenga en cuenta las restricciones de la variable de entrada.

```python
def fibonacci(n):
    if n <= 0:
        return "Error en el número"
    a, b = 0, 1
    serie = []
    for _ in range(n):
        serie.append(a)
        a, b = b, a + b
    return serie

n=input("¿Cuántos términos de Fibonacci quieres?")

if n.isdigit():
  n=int(n)
  resultado = fibonacci(n)
  print(resultado)
else:
  print("Entrada inválida: debe ser un entero positivo")
```

**Ejercicio 9.- Mini-catálogo de sismos.**

En este ejercicio se muestran 4 listas que representan un conjunto de sismos con la siguiente información (fecha, zona, magnitud y profundidad). Hacer un programa que haga lo siguiente:

1.- Convierte la lista de magnitues de str a float.

2.- Los nombres de los estados deben iniciar con mayúscula (utilice una función lambda).

3.- Crea una lista llamada intensidad (con la misma longitud de la lista de magnitudes) y coloca etiquetas de acuerdo al siguiente criterio:

*   Magnitud < 3.5 "Muy Leve"
*   3.5 <= Magnitud < 4.5 "Leve"
*   4.5 <= Magnitud < 3.5 "Moderado"
*   Magnitud > 5.5 "Fuerte"

4.- Con un ciclo For (y condicionales si es necesario), genera un diccionario "conteo_por_zona" con el número de eventos por zona.

5.- Ordena los eventos por magnitud descendente y muestra los 3 sismos más grandes con el siguiente formato:

  YYYY-MM-DD  |  Zona (Estado)  |  Magnitud   |  Profundidad (km).

**Hint:** Recuerde las funciones predefinidas de python map, filter, len, entre otras.

```python
####Datos
Fechas = ["2025-03-01","2025-03-02","2025-03-03","2025-03-04","2025-03-05",
    "2025-03-06","2025-03-07","2025-03-08","2025-03-09","2025-03-10"]

Zonas = ["guerrero","oaxaca","chiapas","michoacán","guerrero",
    "oaxaca","chiapas","guerrero","oaxaca","chiapas"]

magnitudes = ["3.8","4.2","5.0","2.9","4.8","3.5","4.1","5.3","4.0","3.0"]

prof_km = [15, 80, 12, 5, 35, 120, 60, 25, 10, 90] #Profundidad en kilómetros.

####1. CONVERTIR LA LISTA DE MAGNITUDES DE STR A FLOAT
magnitudes_f = list(map(float, magnitudes))

####2. LOS NOMBRES DE LOS ESTADOS DEBEN INICIAR CON MAYÚSCULA
estados = list(map(lambda s: s.capitalize(), Zonas))

####3. INTENSIDAD
clasificar_intensidad = lambda m: (
    "Muy Leve" if m < 3.5 else (
        "Leve" if 3.5 <= m < 4.5 else (
            "Moderado" if 4.5 <= m < 5.5 else "Fuerte"
        )
    )
)
intensidad = list(map(clasificar_intensidad, magnitudes_f))

####4. CONTEO POR ZONA CON EL NÚMERO DE EVENTOS POR ZONA
conteo_por_zona = {}
for zona in estados:
    conteo_por_zona[zona] = conteo_por_zona.get(zona, 0) + 1

####5. ORDENAR LOS EVENTOS POR MAGNITUD DESCENDENTE Y MOSTRAR LOS 3 MÁS GRANDES
eventos = list(zip(Fechas, estados, magnitudes_f, prof_km, intensidad))
tres_mas_grandes = sorted(eventos, key=lambda e: e[2], reverse=True)[:3]

print("Conteo por zona:")
for zona, conteo in conteo_por_zona.items():
    print(f"{zona}: {conteo} eventos")

print("\n3 más grandes:")
for fecha, zona, magnitud, profundidad, intensidad in tres_mas_grandes:
    print(f"{fecha} | {zona} | {magnitud:.1f} | {profundidad} km|({intensidad})")
```

**Ejercicio 10.- Operaciones vectoriales.**

Dadas las siguientes listas:

a = [2, -1, 4]

b = [4, -2, 8]

c = [-5, 2, 1]

Calcule el producto escalar:

1) ab

2) bc

3) ca

Calcular el productor vectorial:

1) a **x** b

2) b **x** c

3) a **x** c

Utilice las listas de inicio para probar su código. Posteriormente, genere una función que reciba como entrada las listas (vectores) y haga las operaciones vectoriales.

Indique si algunos de los vectores son paralelos o perpendiculares.

```python
####Vectores
a = [2, -1, 4]
b = [4, -2, 8]
c = [-5, 2, 1]

####Producto escalar (dot)
def dot(v, w):
  """Producto escalar de dos vectores"""
  return v[0] * w[0] + v[1]*w[1]+ v[2]*w[2]

####Producto vectorial (cross)
def cross(v, w):
  """Producto vectorial de dos vectores"""
  return [v[1] * w[2] - v[2] * w[1],
          v[2] * w[0] - v[0] * w[2],
          v[0] * w[1] - v[1] * w[0]]

#### Calcular Productos escalares
ab = dot(a, b)
bc = dot(b, c)
ca = dot(c, a)

#### Calcular Productos vectoriales
axb = cross(a, b)
bxc = cross(b, c)
cxa = cross(c, a)

print("\nProductos escalares:")
print("a.b =", ab)
print("b.c =", bc)
print("c.a =", ca)

print("\nProductos vectoriales:")
print("a x b =", axb)
print("b x c =", bxc)
print("c x a =", cxa)

############Determinar paralelos (cross=[0,0,0])
def es_paralelo(v, w):
    return cross(v, w) == [0, 0, 0]

############Determinar perpendiculares (dot=0)
def es_perpendicular(v, w):
    return dot(v, w) == 0

print("\nRelaciones:")
if es_paralelo(a, b):
    print("a y b son paralelos")
if es_paralelo(b, c):
    print("b y c son paralelos")
if es_paralelo(c, a):
    print("c y a son paralelos")

if es_perpendicular(a, b):
    print("a y b son perpendiculares")
if es_perpendicular(b, c):
    print("b y c son perpendiculares")
if es_perpendicular(c, a):
    print("c y a son perpendiculares")
```
