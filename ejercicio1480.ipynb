```python
def runningSum(nums):
    n = len(nums)
    prefix_sum = [0] * (n + 1)
    for i in range(n):
        prefix_sum[i + 1] = prefix_sum[i] + nums[i]
    return prefix_sum

def range_sum(prefix_sum, left, right):
    return prefix_sum[right + 1] - prefix_sum[left]

# Ejemplo
nums = [1,2,3,4]
prefix = runningSum(nums)
print(f"Ejemplo original 1: {nums}")
print(f"Prefix sum array: {prefix}")
nums = [1,1,1,1,1]
prefix = runningSum(nums)
print(f"Ejemplo original 2: {nums}")
print(f"Prefix sum array: {prefix}")
nums = [3,1,2,10,1]
prefix = runningSum(nums)
print(f"Ejemplo original 3: {nums}")
print(f"Prefix sum array: {prefix}")
```

```python
class Solution:
    def runningSum(self, nums: list[int]) -> list[int]:
        n = len(nums)
        ans = [0] * n
        for i in range(n):
            current_sum = 0
            for j in range(i + 1):
                current_sum += nums[j]
            ans[i] = current_sum
        return ans

# Example
sol = Solution()
ejemplo_nums = [1, 2, 3, 4]
result = sol.runningSum(ejemplo_nums)
print(f"Original array 1: {ejemplo_nums}")
print(f"Running Sum 1: {result}")

ejemplo_nums_2 = [1,1,1,1,1]
result_2 = sol.runningSum(ejemplo_nums_2)
print(f"Original array 2: {ejemplo_nums_2}")
print(f"Running Sum 2: {result_2}")

ejemplo_nums_3 = [3,1,2,10,1]
result_3 = sol.runningSum(ejemplo_nums_3)
print(f"Original array 3: {ejemplo_nums_3}")
print(f"Running Sum 3: {result_3}")
```
